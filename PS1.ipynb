```python
#1.Flowchart

def Print_value(a,b,c):
    if a>b:
        if b>c:
            print(a,b,c)
        else:
            if a>c:
                print(a,c,b)
            else:
                print(c,a,b)
    else:
        if b>c:
            return
        else:
          print(c,b,a)

import random
Print_value(250,200,50)

a=random.randint(1,50)
b=random.randint(1,50)
c=random.randint(1,50)
print(a,b,c)
Print_value(a,b,c)
```

```python
#3. Pascal triangle
def Pascal_triangle(k):
    if k == 1:
        return[1]
    row=[1]
    for i in range(1,k):
        roww=[1]
        for j in range(1,i):
            roww.append(row[j-1]+row[j])
        roww.append(1)
        row=roww
    return row

print(Pascal_triangle(100))
print(Pascal_triangle(200))
```

```python
#4. Add or double
def Least_moves(x):
    moves = 0
    while x>1:
        if x%2==0:
            x=x/2  
        else:
            x=x-1  
        moves =moves+1
    return moves


print( Least_moves(4))
print(Least_moves(7))

x1= random.randint(1, 100)
print(x1)
print( Least_moves(x1))
```

```python
#5.1 Dynamic programming
def Find_expression(y):
    expressions =["1"]
    for num in range(2,10):
        expressions2=[]
        for j in expressions:
            plus=j+"+"+str(num)
            expressions2.append(plus)
            minus=j+"-"+str(num)
            expressions2.append(minus)
            connect=j+str(num)
            expressions2.append(connect)
        expressions=expressions2
         

    for j in expressions:
      value=eval(j)
      if value== y:
        print(j,"=",y)

      
Find_expression(50)   


    
```

```python
expressions = ["1"]
for num in range(2, 10):
    expressions2=[]
    for j in expressions:
        plus=j+"+"+str(num)
        expressions2.append(plus)
        minus=j+"-"+str(num)
        expressions2.append(minus)
        connect=j+str(num)
        expressions2.append(connect)
    expressions=expressions2

#5.2
dictionary={}
for j in expressions:
    result=eval(j) 
    if result not in dictionary:
        dictionary[result]=0
    dictionary[result]=dictionary[result]+1

Total_solutions=[]
for i in range(1, 101):
    count = 0
    if i in dictionary:
        count=dictionary[i]
    Total_solutions.append(count)

maxcount=max(Total_solutions)
mincount=min(Total_solutions)
maxnumber=Total_solutions.index(maxcount) + 1
minnumber=Total_solutions.index(mincount) + 1

print(maxnumber, maxcount)
print(minnumber, mincount)

import matplotlib.pyplot as plt
number=range(1, 101) 
plt.plot(number,Total_solutions)
plt.xlabel("Target Number (1-100)")
plt.ylabel("Number of Solutions")
plt.title("Total solutions for each target number")
plt.show
```
